```python
from pathlib import Path
R=Path('/kaggle/input/datasets')
print('utilisateur:', sorted(p.name for p in R.iterdir()))
for u in sorted(R.iterdir()):
    for d in sorted(u.iterdir()):
        n=sum(1 for _ in d.rglob('*'))
        print(f'  {u.name}/{d.name}  entrees={n}')
        for q in sorted(d.iterdir())[:5]:
            print('      ', q.name)
```
